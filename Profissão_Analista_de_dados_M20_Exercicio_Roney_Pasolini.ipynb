### **1.1. Nativo**

Neste primeira atividade, você deve ler o código abaixo implementado em Python nativo e responder a seguinte pergunta:

> O que o código abaixo computa?

**Resposta**: O código abaixo computa os valores de todos os incidentes que aconteceram em determinado dia.

## Explicação do código

Primeiro ele lê o arquivo, e salva todas os dados na variavel data.
Ele configura as variaveis dia como 14 (pois é o dia em que os dados começaram a ser projetados) e incidents como 0. Cria um dicionario vazio com o nome Incidente_by_day.

No for principal ele trata cada linha do csv de forma separada (cada loop é uma linha)

Em cada loop principal, ele separa os valores das linhas em uma lista chamada timebox_data. Após a separaçaõ, ele cria um loop secundario para somar cada um dos dadaos da linha, para ir somento o numero de incidentes na variavel incidents.

Após realizar a somatoria de incidentes de uma linha, ele captura o valor da primeira coluna, que se refere a hora. cuja o valor varia entre 1 e 27. (Devido a como o csv foi configurado, ele computa um valor inteiro a cada 30 minutos. Lembrando que os dados foram capturados de 07 as 20, o que resulta em 26 horarios. **IMPORTANTE: A hora é computada a partir de 1, devido a isso o valor maximo é 27**)

Caso o valor seja 27, ele salva o numero do dia e o numero de incidentess no dicionario, aumenta o contador de dia em 1 e retorna o valor de incidentes para 0. Recomeçando o loop principal

### Explicação da saida

O código imprime todo o dicionario de incident_by_day



```
# -- read

data = None

with open(file='traffic.csv', mode='r', encoding='utf8') as fp:

  fp.readline()
  data = fp.read()
  #print(data)

# -- analytics

day = 14
incidents = 0
incident_by_day = dict()

print(data.split(sep='\n')[14])

for timebox in data.split(sep='\n'): # Esta separando os dados de cada linha do csv

  timebox_data = timebox.split(sep=';') # Está separando os dados de cada linha

  # --
  # -- inicio da computação escalar
  # --

  for incident in timebox_data[1: len(timebox_data)-1]: # Esta verificando a existencia de valores acima de 0. Ignorando a primeira coluna (hora).
    incidents = incidents + int(incident)
    #print(incident)
  # --
  # -- fim da computação escalar
  # --

  try:

    half_hour = int(timebox_data[0]) #Captura o valor da hora e verifica se é 27. Correspondendo ao ultimo horario de um dia, se for elle soma o dia e reseta o numero de incidentes para uma nova contagem.

    if half_hour == 27:
      incident_by_day[day] = incidents # Savla os valores de dia e numero de incidentes em um dicionario python
      day = day + 1
      incidents = 0

  except ValueError:
    continue

# -- results

for day in incident_by_day:

  print(f'{day}: {incident_by_day[day]}')
```

### **1.2. NumPy**

Substitua o trecho do código do algoritmo que utiliza da **computação escalar** por um que utiliza da **computação vetorial**. Use o pacote NumPy.

```
# -- read

data = None

with open(file='traffic.csv', mode='r', encoding='utf8') as fp:

  fp.readline()
  data = fp.read()

  #print(data)

# -- analytics

import numpy as np

day = 14
incidents = 0
incident_by_day = dict()

ONE_VALUE = np.ones(16, dtype=int)

for timebox in data.split(sep='\n')[0:-1]: #Verificado que o CSV possui uma linha adicional em branco

  timebox_data = timebox.split(sep=';')


  timebox_array = np.array(timebox_data[1:-1] , dtype=int)

  # --
  # -- inicio da computação vetorial
  # --


  incidents = incidents + np.dot(timebox_array , ONE_VALUE)

  # --
  # -- fim da computação vetorial
  # --

  try:

    half_hour = int(timebox_data[0])

    if half_hour == 27:
      incident_by_day[day] = incidents
      day = day + 1
      incidents = 0

  except ValueError:
    continue

# -- results


for day in incident_by_day:

  print(f'{day}: {incident_by_day[day]}')
```
